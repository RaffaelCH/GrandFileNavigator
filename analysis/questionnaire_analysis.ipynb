# Questionnaire Analysis

Code to analyze the questionnaire responses: calculate metrics, match them to tasks.

```python
from shared import *
import pandas


def map_agreement(agreement):
    if agreement == "Strongly agree":
        return 5
    elif agreement == "Agree":
        return 4
    elif agreement == "Neutral":
        return 3
    elif agreement == "Disagree":
        return 2
    elif agreement == "Strongly disagree":
        return 1
    return agreement

def drop_unneeded(dataframe):
    return dataframe.drop(["Start time", "Completion time", "Email", "Name", "Last modified time"], axis=1)


def preprocess(dataframe):
    dataframe = drop_unneeded(dataframe)
    dataframe = dataframe.map(map_agreement)
    return dataframe

evaluation_data = pandas.read_csv("./questionnaire_data/Evaluation_Feedback.csv", sep=";")
evaluation_data = preprocess(evaluation_data)
overall_data = pandas.read_csv("./questionnaire_data/Overall_Feedback.csv", sep=";")
overall_data = preprocess(overall_data)
```

```python
excluded_participant_id = "P2" # strong outlier -> ignore

filtered_evaluation_data = evaluation_data[evaluation_data["ID"] != excluded_participant_id]
filtered_overall_data = overall_data[overall_data["ID"] != excluded_participant_id]
```

```python
task_one_participant_ids = [participant.name for participant in participants if participant.evalTask == EvaluationTask.T1]
task_two_participant_ids = [participant.name for participant in participants if participant.evalTask == EvaluationTask.T2]

evaluation_task_one = filtered_evaluation_data[filtered_evaluation_data.apply(lambda row: row["ID"] in task_one_participant_ids, axis=1)]
evaluation_task_two = filtered_evaluation_data[filtered_evaluation_data.apply(lambda row: row["ID"] in task_two_participant_ids, axis=1)]
overall_task_one = filtered_overall_data[filtered_overall_data.apply(lambda row: row["ID"] in task_one_participant_ids, axis=1)]
overall_task_two = filtered_overall_data[filtered_overall_data.apply(lambda row: row["ID"] in task_two_participant_ids, axis=1)]
```
